#Importing all the necessary libraries
**#Overall steps involved are:**
1. Reading the image as 3d array .
2. Reshaping it into 2d array.
3. Using kmeans to find the required number of color clusters in the image.
4. Rounding the centers of all the clusters.
5. Using the model to predict which points lie in which cluster.
6. Reshaping the array in order to view it as an image .

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

```python
img_arr_3d= mpimg.imread('palm_trees.jpg')
```

```python
img_arr_3d.shape
```

```python
plt.imshow(img_arr_3d)
```

```python
h,w,c=img_arr_3d.shape
```

```python
# converting form 3d to 2d
img_arr_2d=img_arr_3d.reshape(h*w,c)
```

# Using Kmeans clustering

```python
from sklearn.cluster import KMeans
model=KMeans(n_clusters=5)
labels=model.fit_predict(img_arr_2d)
```

```python
rgb_codes=model.cluster_centers_.round(0).astype('int')
```

```python
p=rgb_codes[labels]
```

```python
final=np.reshape(p,(h,w,c))
```

#Output of given number of colors in the final image separately

```python
plt.imshow(final)
```
